# Калькулятор

```python
def calculator():
    print("Простой калькулятор")
    print("Операции: +, -, *, /")
    
    a = float(input("Первое число: "))
    op = input("Операция: ")
    b = float(input("Второе число: "))
    
    if op == "+":
        result = a + b
    elif op == "-":
        result = a - b
    elif op == "*":
        result = a * b
    elif op == "/":
        if b == 0:
            print("На ноль делить нельзя!")
            return
        result = a / b
    else:
        print("Неизвестная операция")
        return
    
    print(f"Результат: {result}")

calculator()
```
